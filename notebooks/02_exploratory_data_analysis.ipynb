### Objective
Analyze patterns in food shelf-life deterioration, investigate correlations, and visualize key features.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('../data/processed/carekart_food_cleaned.csv')
print(f'Loaded dataset: {df.shape[0]} rows, {df.shape[1]} columns')
df.info()
```

### 1. Target Variable (Remaining Shelf-Life) Distribution

```python
plt.figure(figsize=(8, 4.5))
plt.hist(df['remaining_shelf_life_hours'], bins=25, color='#10B981', edgecolor='#065F46')
plt.title('Distribution of Remaining Shelf-Life Hours')
plt.xlabel('Remaining Hours')
plt.ylabel('Count')
plt.grid(True, alpha=0.3)
plt.show()
```

### 2. Shelf-Life Variation by Food Category

```python
cat_summary = df.groupby('category')['remaining_shelf_life_hours'].mean().sort_values()
plt.figure(figsize=(9, 5))
cat_summary.plot(kind='barh', color='#059669')
plt.title('Mean Shelf-Life by Category (Hours)')
plt.xlabel('Hours')
plt.grid(axis='x', alpha=0.3)
plt.show()
```

### 3. Correlation Heatmap

```python
num_cols = ['temperature_c', 'humidity_percent', 'hours_since_cooking', 'moisture_level', 'remaining_shelf_life_hours', 'safety_score']
corr = df[num_cols].corr()
print('Correlation Matrix:')
print(corr.round(3))
```
